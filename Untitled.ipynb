```python
# Caso 1: Índices idénticos (Componente a componente)
serie1 = pd.Series([1, 2], index=['X', 'X'])
serie2 = pd.Series([10, 20], index=['X', 'X'])
serie1 + serie2

```

```python
# Caso 2: Índices diferentes con duplicados (Producto cartesiano)
serie3 = pd.Series([1, 2], index=['X', 'Y'])
serie4 = pd.Series([10, 20], index=['X', 'X'])
serie3 + serie4
```
